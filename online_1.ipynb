```python
import pandas as pd # 導入 pandas 庫並將其命名為 pd
df1 = pd.read_csv('/content/drive/MyDrive/Online/online_retail.csv') # 從指定路徑讀取 CSV 檔案並存儲到 df1 DataFrame 中
```

### 檢查缺失值

首先，我將檢查 `df1` DataFrame 中是否存在任何缺失值。

```python
print(df1.isnull().sum()) # 計算並列印 df1 中每個欄位的缺失值數量
```

### 處理缺失值

現在，我將刪除包含任何缺失值的列，以確保數據集的乾淨。

```python
df_cleaned = df1.dropna() # 刪除 df1 中所有含有缺失值的行，並將結果存儲到 df_cleaned
print(df_cleaned.isnull().sum()) # 再次檢查 df_cleaned 中每個欄位的缺失值數量
```

### 儲存清理後的資料
### 資料不包含空白的值

最後，我將把清理後的 DataFrame 儲存為一個名為 `online_retail_cleaned.csv` 的新 CSV 檔案，並放置在 `/content/drive/MyDrive/Online/` 目錄中。

```python
output_path = '/content/drive/MyDrive/Online/online_retail_cleaned.csv' # 定義輸出檔案的路徑
df_cleaned.to_csv(output_path, index=False) # 將清理後的 df_cleaned 儲存為 CSV 檔案，不包含索引
print(f"Cleaned data saved to: {output_path}") # 列印清理後的數據儲存路徑
```
